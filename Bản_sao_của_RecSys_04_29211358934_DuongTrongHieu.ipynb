## Situation 3 - Content-based filtering  *(medium)*

**Studio problem.** Popularity gives everyone the same list. Content-based
filtering personalises with **item features**. First build **Vy's taste
profile**: a vector of 4 numbers, one per feature (`seafood`, `spicy`, `noodle`, `vegetarian`).

```
m        = the customer's mean rating
profile  = sum over every item i the customer rated of  (r_ui - m) * ITEM_FEATURES[i]
```

Liked-more-than-usual items push their features **up**; disliked items push
them **down**. Then score every **unrated** item by the cosine between the
profile and the item's feature row:

```
cosine(a, b) = a . b / (|a| |b|)          (return 0.0 if either vector is all zeros)
```

Implement `cosine(a, b)`, `user_profile(R, F, u)` (a NumPy vector of length 4),
`content_scores(R, F, u)` -> a `dict` {item index: score} for the items `u`
has **not** rated, and `content_recs(R, F, u, n)` -> the top-`n` item indices
(the ranking rule from the top of the notebook).

Then store:

- `S3_profile` - Vy's profile as a list of 4 floats, each rounded to 4
- `S3_liked_feature` - the **name** of the feature with the largest profile value
- `S3_top3` - the 3 recommended recipe **names**; `S3_top_score` - the best score (4 decimals)
- `S3_comment` - **2-3 sentences**: which features does *your* Vy like and
  dislike, and how do they explain the top-3? Name one recipe content-based
  can never recommend to Vy and say why (the *filter bubble*).

````python
def cosine(a, b):
    a = np.asarray(a, dtype=float)
    b = np.asarray(b, dtype=float)
    na, nb = np.linalg.norm(a), np.linalg.norm(b)
    if na == 0.0 or nb == 0.0:
        return 0.0
    return float(np.dot(a, b) / (na * nb))

def user_profile(R, F, u):
    rated = ~np.isnan(R[u])
    if not np.any(rated):
        return np.zeros(F.shape[1], dtype=float)
    mean_u = np.nanmean(R[u])
    diffs = R[u, rated] - mean_u
    return np.dot(diffs, F[rated])

def content_scores(R, F, u):
    p = user_profile(R, F, u)
    return {i: cosine(p, F[i]) for i in range(R.shape[1]) if np.isnan(R[u, i])}

def content_recs(R, F, u, n):
    sc = content_scores(R, F, u)
    return sorted(sc, key=lambda i: (-sc[i], i))[:n]

_vy_prof = user_profile(R, ITEM_FEATURES, TARGET)
S3_profile = [round(float(x), 4) for x in _vy_prof]
S3_liked_feature = FEATURES[int(np.argmax(_vy_prof))]
_c_top = content_recs(R, ITEM_FEATURES, TARGET, TOP_N)
S3_top3 = [ITEMS[i] for i in _c_top]
S3_top_score = round(float(content_scores(R, ITEM_FEATURES, TARGET)[_c_top[0]]), 4)

S3_comment = (
    "Vy strongly likes spicy dishes (+2.8) and mildly likes seafood (+0.2), while she actively dislikes noodle dishes (-2.2) "
    "after rating Mi Quang, Bun Cha Ca, and Fish Cake Soup at 1 to 2 stars, which explains why spicy Tofu Lemongrass and Crab Glass Noodles "
    "rise to the top of her recommendations. However, content-based filtering creates a filter bubble and can never recommend "
    "Morning Glory or Veggie Spring Rolls to Vy because their feature vectors are purely vegetarian [0, 0, 0, 1], giving a dot product and cosine score of 0.0."
)

print("profile", dict(zip(FEATURES, S3_profile or [])), "| likes most:", S3_liked_feature)
print("content top-3 for", TARGET_USER, ":", S3_top3, "| best score", S3_top_score)
````
